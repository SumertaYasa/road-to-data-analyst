### Tipe Data

Tipe Data adalah sebuah memori yang digunakan untuk menyimpan/menampun data.

Tipe Data di Python terbagi menjadi dua jenis:
- Tipe Data Primitif
- Tipe Data Non-primitif

#### Tipe Data Primitif
Tipe data ini hanya dapat menyimpan satu nilai tunggal dalam satu variabel dan merupakan tipe data dasar bawaan dari Python.
* **`int`**: Bilangan bulat (contoh: `10`, `-5`).
* **`float`**: Bilangan desimal (contoh: `3.14`, `2.0`).
* **`str`**: Teks atau kumpulan karakter (contoh: `"Aditya"`, `"2022"`).
* **`bool`**: Nilai logika kebenaran (contoh: `True`, `False`).
* **`None`**: Representasi nilai kosong atau tidak ada nilai sama sekali.

#### Tipe Data Non-primitif
Tipe data ini dapat menampung lebih dari satu nilai dalam satu variabel dan biasanya memikiki struktur tertentu.
* **`list`**: Kumpulan data yang berurutan dan bisa diubah isinya (*mutable*). (contoh: `[1, 2, 3]`).
* **`tuple`**: Kumpulan data yang berurutan namun tidak bisa diubah isinya (*immutable*). (contoh: `(10, 20)`).
* **`dict` (Dictionary)**: Kumpulan data dalam bentuk pasangan kunci dan nilai (*key-value pair*). (contoh: `{"nama": "Adi", "skill": "Python"}`).
* **`set`**: Kumpulan data unik yang tidak berurutan dan tidak mengizinkan duplikasi. (contoh: `{1, 2, 3}`).

```python
# TIPE DATA PRIMITIF

print("Tipe Data Primitif:\n")
# string
print("Hello World")
# integer
print(1)
# float
print(1.1)
# boolean
print(True)

# TIPE DATA NON-PRIMITIF
print("\n")
print("Tipe Data Non-Primitif:\n")
# list -> simple-nya array
print([1,2,3,4,5])
print(["satu", "dua", "tiga"])
# tuple -> simple-nya array konstanta
print((1,2,3,4,5))
print(("satu", "dua", "tiga"))

# dictionary -> simple-nya object
print({"name": "John", "age": 30})
```
